**a. How many rows and columns are contained in the dataset?**

```python
import pandas as pd
import numpy as np

df = pd.read_csv('poll-responses.csv')

print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

**b. What datatypes are contained in the dataset?**

```python
print(df.dtypes)
```

**c. Is the dataset complete? i.e. no missing values?**

```python
print(f"Missing values per column:\n{df.isnull().sum()}")
```

**d. Slice out the first 15 rows and last 20 rows and merge them into a new
dataframe titled `df_sample`. Display this new dataframe.**

```python
first_15 = df.head(15)
last_20  = df.tail(20)

df_sample = pd.concat([first_15, last_20], ignore_index=True)

display(df_sample)
```
